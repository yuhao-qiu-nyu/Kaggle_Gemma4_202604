```python
import json

# 请把这 22 个词替换成你实际的单词列表
your_words = [
    "bye", "please", "thank_you", "sorry", "excuse_me", 
    "help", "what", "who", "where", "why",
    "no", "yes", "hi", "eat", "you",
    "drink", "how", "know", "name", "friend",
    "me", "toilet"
]

# 自动生成字典
sign_map = {word.lower(): i for i, word in enumerate(your_words)}

# 写入文件（这会自动处理好所有逗号）
with open('sign_map.json', 'w', encoding='utf-8') as f:
    json.dump(sign_map, f, indent=4)

print("✅ sign_map.json 已重新生成，格式现在是完美的！")
```
